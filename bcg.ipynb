```python
def sommepair(l):
    acc=0
    for x in l:
        if x%2 == 0:
            acc+=x
    return acc
```

```python
sommepair([2,4])
sommepair([1,3])
```

```python
def maxim(l):
    imax = l[0]
    for x in l:
        if x>imax:
            imax = x
    return imax
```

```python
maxim([1,2,5,2,3])
```

```python
def estPalindrome(str):
    if str == str[::-1]:
        return True
    else:
        return False
```

```python
estPalindrome('radar')
estPalindrome('radat')
```

```python
def premier_doublon(lst):
    vus = []
    for x in lst:
        if x in vus:
            return x
        else:
            vus.append(x)
    return None
```

```python
premier_doublon([1,2,3])
```

```python
def chart_frequent(lst):
    acc1 =0
    bestcount =0
    vus=[]
    for x in lst:
        if x in vus:
            acc1+=1
        else : 
            vus.append(x)
        if acc1>bestcount :
            bestcount=acc1
    return bestcount
```

```python
chart_frequent('viande')
```

```python
def elementcommun(lst1,lst2):
    lstcomm = []
    for x in lst1:
        for y in lst2:
            if x==y:
                lstcomm.append(x)
    return lstcomm
```

```python
elementcommun([1,2,3],[2])
```

```python
def estAnagram(s, t):
    if len(s) != len(t):
        return False

    # Étape 1 : construire compteur pour s
    compteur_s = {}
    for c in s:
        compteur_s[c] = compteur_s.get(c, 0) + 1

    # Étape 2 : construire compteur pour t
    compteur_t = {}
    for c in t:
        compteur_t[c] = compteur_t.get(c,0)+1   # même logique

    # Étape 3 : comparer les deux dictionnaires
    return compteur_s == compteur_t
```

```python
estAnagram("anagram","aragnan")
```

```python
def excepti(nums):
    n = len(nums)
    answer = [1] * n

    # Passe 1 : produits de gauche
    gauche = 1
    for i in range(n):
        answer[i] = gauche
        gauche *= nums[i]

    # Passe 2 : multiplier par produits de droite
    droite = 1
    for i in range(n-1, -1, -1):
        answer[i] *= droite
        droite *= nums[i]

    return answer

        
```

```python
excepti([1,2,3,4,5,6])
```

```python
def maxSubArray(nums, k):
    # Somme de la première fenêtre
    window = sum(nums[:k])
    best = window

    # Faire glisser la fenêtre
    for i in range(k, len(nums)):
        window = window - nums[i-k] + nums[i]  # enlève gauche, ajoute droite
        best = max(best, window)

    return best
```

```python
maxSubArray([1,2,3,4,5,1,2],2000)
```

```python
def estferme(s):
    stack=[]
    pairs = {
        ')':'(',
        ']':'[',
        '}':'{'
    }
    for c in s:
        if c in '([{':
            stack.append(c)
        else:
            if not stack:
                return False
            elif stack[-1]!=pairs[c]:
                return False
            stack.pop()
    return len(stack)==0
```

```python
estferme("([)")
```
